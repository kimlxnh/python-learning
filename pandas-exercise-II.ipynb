# Pandas Exercise II

```python
import pandas as pd
```

First, we will load the dataset from `AAPL.csv` into a DataFrame.

```python
df = pd.read_csv('./AAPL.csv')
df.head()
```

This data, in its raw format, is the same as that which can be retrieved from a number of financial websites.

This dataset has 5 years of Apple (AAPL) share prices, one row per trading day, from 30 June 2015 to 29 June 2020:

- `Open` / `Close` – the price when the market opened and closed that day
- `High` / `Low` – the highest and lowest price that day
- `Adj Close` – the closing price adjusted for events such as share splits, so prices from different years can be compared fairly
- `Volume` – how many shares were traded that day

Before starting the exercise, let's add some additional data columns, calculated from the raw data. Don't worry if you aren't familiar with the methods used in the following cell.

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.day_name()
df['Change %'] = (df['Adj Close'].pct_change() * 100)
```

## Combining conditions

#### 15. How many days in March 2020 did the price go down (`Change %` below 0)?

Use `&` to combine the conditions, and store the answer in a variable called `down_days_march_2020`

*Hint: remember to put each condition inside its own round brackets*

```python
# Add your code here
down_days_march_2020 = ((df_2020['Month']==3) & (df_2020['Change %'] < 0)).sum()
print(down_days_march_2020)
```

#### 16. How many days did the price move more than 5% either way (`Change %` above 5 or below -5)?

Use `|` to combine the conditions, and store the answer in a variable called `big_move_days`

*Hint: save the mask in a variable called `big_move_msk` – you'll need it in question 17*

```python
# Add your code here msk is a series, and days stores total no. of big moves
big_move_msk = ((df['Change %'] > 5) | (df['Change %'] < -5))
big_move_days = big_move_msk.sum()
print(big_move_days)
```

#### 17. Use `.loc` to display only the `Date`, `Adj Close` and `Change %` columns for the days in question 16

```python
# Add your code here row msk and columns date adj close and change
df.loc[big_move_msk, ['Date', 'Adj Close', 'Change %']]
```
