```python
df = df.drop(columns='Ticket')
df.head()
```

```python
df.columns = df.columns.str.lower()
df.head()
```

```python
df = df.rename(columns={'pclass': 'passenger_class'})
df.head()
```

```python
df['survived'].value_counts()
```

```python
df['survived'] = df['survived'].astype(bool)
df['survived'].value_counts()
```

```python
df['embarked'].value_counts()
```

```python
df['embarked'] = df['embarked'].map({'S':'Southampton', 'C':'Cherbourg', 'Q': 'Queenstown'})
df['embarked'].value_counts()
```

```python
df['relatives'] = df['sibsp'] + df['parch']
df.head()
```

```python
#df = df.drop(columns=['sibsp', 'parch'])
df.head()
```

```python
name_split = df['name'].str.split(', ', n=1, expand=True, regex=False)
name_split.head()
```

```python
df['surname'] = name_split[0]
df.head()
```

```python
title_split = name_split[1].str.split('. ', n=1, expand=True, regex=False)
title_split.head()
```

```python
df['title']= title_split[0]
df['first_name'] = title_split[1]
df.head()
```

```python
df['title'].value_counts()
```

```python
df = df.drop(columns='name')
df.head()
```

```python
df['first_name'].str.contains('(', regex=False).sum()
```

```python
extracted = df['first_name'].str.extract(r'\((.*?)\)')[0]
extracted
```

```python
df['first_name'] = extracted.fillna(df['first_name'])
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns='cabin')
df.head()
```

```python
df['embarked'].value_counts()
```

```python
df['embarked'] = df['embarked'].fillna('Southampton')
df['embarked'].isnull().sum()
```

```python
df.groupby('title')['age'].median()
```

```python
df['age'] = df['age'].fillna(df.groupby('title')['age'].transform('median'))
print(df['age'].isnull().sum())
```

```python
df.head()
```

```python
df = df[['passengerid', 'title', 'first_name', 'surname', 'survived', 'passenger_class', 'sex', 'age', 'relatives', 'fare', 'embarked']]
df.head()
```

```python
df = df.rename(columns={'passengerid': 'passenger_id', 'sex': 'gender'})
df.head()
```

#### Who was most likely to survive?

```python
# females in 3rd class, rounded to 1 d.p
print(df.loc[third_class_mask & (df['gender']=='female'), 'survived'].mean().round(2))
```

```python
#females in 1st class
print(df.loc[(df['passenger_class'] == 1) & (df['gender']=='female'), 'survived'].mean().round(2))
```

```python
#children survival chances
print(df.loc[df['age']<18, 'survived'].mean().round(2))
```

```python
# adults survival chances
print(df.loc[df['age']>=18, 'survived'].mean().round(2))
```
