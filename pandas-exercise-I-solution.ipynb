#### Stretch: Use `loc` to select the rows where `carat` is greater than 0.3 and display the `carat`, `cut`, and `price` columns

*Try searching: "pandas boolean indexing"*

```python
# Add your code here

df.loc[df['carat'] > 0.3, ['carat', 'cut', 'price']]
```

#### Stretch: Create a copy of the DataFrame and add a new column `expensive` which is `True` if price is greater than 1000, otherwise `False`

*Try searching: "pandas new column based on condition"*

```python
# Add your code here

df2 = df.copy()
df2['expensive'] = df2['price'] > 1000

df2.sample(15)
```

#### Stretch: Sort the DataFrame by the `price` column in ascending order, and `carat` in descending order

*Try searching: "pandas sort_values multiple columns"*

```python
# Add your code here

df2.sort_values(by=['price', 'carat'], ascending=[True, False])
```

#### Stretch: Use `loc` to filter the DataFrame for diamonds that have a `carat` greater than 0.5 and a `price` less than 1500. Display the `carat`, `price`, and `cut` columns

*Try searching: "pandas filter multiple conditions"*

```python
# Add your code here

df2.loc[(df2['carat'] > 0.5) & (df2['price'] < 1500), ['carat', 'price', 'cut']]
```

#### Stretch: Replace all instances of '`Good`' in the `cut` column with '`Average`'

*Try searching: "pandas replace values in column"*

```python
# Add your code here

df2['cut'] = df2['cut'].replace('Good', 'Average')
df2.sample(10)
```

#### Group the DataFrame by the `cut` column and calculate the average `price` for each group

```python
# Add your code here

df2.groupby('cut')['price'].mean()
```

#### Stretch: Rename the columns `x`, `y`, and `z` to `length`, `width`, and `height`, respectively

*Try searching: "pandas rename columns"*

```python
# Add your code here

df2.rename(columns={'x': 'length',
                    'y': 'width',
                    'z': 'height'}
        ).head(10)
```
